```python
import os
import subprocess
import sys
from importlib.util import find_spec

try:
    ROOT = os.path.dirname(os.path.abspath(__file__))
except NameError:
    ROOT = os.getcwd()  # fallback for Jupyter

REQUIRED_MODULES = [
    'ipywidgets',
    'jupyterlab_widgets',
    'widgetsnbextension',
    'openpyxl',
    'pandas',
    'requests'
]

def install_requirements():
    missing = [m for m in REQUIRED_MODULES if find_spec(m) is None]

    if missing:
        print(f"Installing missing modules: {missing}")
        subprocess.check_call([
            sys.executable, '-m', 'pip', 'install',
            '-r', os.path.join(ROOT, 'requirements.txt')
        ])
    else:
        print("All required modules already installed")

def main():
    # Ensure project root is in path
    if ROOT not in sys.path:
        sys.path.insert(0, ROOT)

    install_requirements()

    # Clear cached modules
    for name in list(sys.modules):
        if name == 'core' or name.startswith('core.'):
            del sys.modules[name]

    print("Setup complete. You can now run your app.")

if __name__ == "__main__":
    main()
```
